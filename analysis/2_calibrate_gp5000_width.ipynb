## Fit a width model

```python
mounted_width_samples = (
    cyclingnews_widths
    .melt(
        id_vars=["wheelset", "axle", "internal_width_mm"],
        value_vars=["str_28_width_mm", "str_30_width_mm"],
        var_name="tire_model",
        value_name="mounted_width_mm",
    )
    .assign(
        nominal_width_mm=lambda dataframe: dataframe["tire_model"]
        .str.extract(r"(\d+)")[0]
        .astype(int)
    )
)
```

```python
mounted_width_samples.head()
```

```python
def fit_width_model(tire_width_samples):
    internal_width_mm = tire_width_samples["internal_width_mm"]
    mounted_width_mm = tire_width_samples["mounted_width_mm"]
    width_slope_per_mm = internal_width_mm.cov(mounted_width_mm) / internal_width_mm.var()
    reference_internal_width_mm = 23.5

    return pd.Series(
        {
            "sample_count": len(tire_width_samples),
            "reference_internal_width_mm": reference_internal_width_mm,
            "mounted_width_at_reference_mm": (
                mounted_width_mm.mean()
                + width_slope_per_mm
                * (reference_internal_width_mm - internal_width_mm.mean())
            ),
            "width_slope_per_mm": width_slope_per_mm,
        }
    )


s_tr_width_calibration = (
    mounted_width_samples
    .groupby("nominal_width_mm", sort=True)
    .apply(fit_width_model, include_groups=False)
    .reset_index()
)
```

```python
s_tr_width_calibration.head()
```

```python
assert s_tr_width_calibration.loc[
    s_tr_width_calibration["nominal_width_mm"] == 28,
    "sample_count",
].item() == 32
assert s_tr_width_calibration.loc[
    s_tr_width_calibration["nominal_width_mm"] == 30,
    "sample_count",
].item() == 32
s_tr_width_calibration.round(3)
```
